# 01 — Load NSL-KDD

**Goal:** install libraries, download the NSL-KDD dataset, load `KDDTrain+` / `KDDTest+` with the official 43-column schema, inspect it (EDA), and save `data/kdd_raw.csv`.

```python
# ============================================================
# 0. Install libraries (run once in Colab)
# ============================================================
%pip install -q pandas numpy scikit-learn scipy joblib matplotlib seaborn requests pennylane

import sys, os
from pathlib import Path

# Make sure the repository root (which contains src/) is on the path.
REPO_ROOT = Path.cwd()
while not (REPO_ROOT / "src").exists() and REPO_ROOT.parent != REPO_ROOT:
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
print("repo root:", REPO_ROOT)
```

```python
# ============================================================
# 1. Download dataset (with mirror fallback)
# ============================================================
from src.loader import download_dataset

paths = download_dataset(REPO_ROOT / "datasets")
paths
```

```python
# ============================================================
# 2. Load KDDTrain+ and KDDTest+
# ============================================================
from src.loader import load_all

df = load_all(REPO_ROOT / "datasets", merge=True)  # adds a `split` column
df.head()
```

```python
# ============================================================
# 3. Inspect dataset (EDA)
# ============================================================
from src.loader import inspect_dataframe

summary = inspect_dataframe(df)
summary
```

```python
# Quick label distribution across train/test splits
import pandas as pd
ct = pd.crosstab(df["split"], df["label"])
ct
```

```python
# ============================================================
# 4. Save cleaned raw dataframe -> data/kdd_raw.csv
#    (the `split` column is dropped so the CSV keeps the exact
#     43-column NSL-KDD schema)
# ============================================================
from src.loader import save_raw

save_raw(df, REPO_ROOT / "data" / "kdd_raw.csv")
```
